## 2.  Indexing, Slicing and Joining

````python
import torch
x = torch.arange(6).view(2,3)
print(x)
````

### Tensor Slicing

Tensor slicing selects part of a tensor.

**Syntax:** `x[rows, columns]`

```python
x = torch.arange(6).view(2, 3)
# tensor([[0, 1, 2],
#         [3, 4, 5]])

x_sub = x[:1, :2]
# tensor([[0, 1]])

````python
x_sub = x[:1,:2]
print(x_sub)
````

### Tensor Index Selection (`index_select`)

`torch.index_select()` selects specific rows or columns using their indices.

**Syntax:**
```python
torch.index_select(input, dim, index)
```

- `input`: The original tensor.
- `dim`: The dimension to select (`0` = rows, `1` = columns).
- `index`: A 1D tensor containing the indices to select.

**Example:**
```python
x = torch.arange(6).view(2, 3)
# tensor([[0, 1, 2],
#         [3, 4, 5]])

# Select row 1
torch.index_select(x, dim=0, index=torch.tensor([1]))
# tensor([[3, 4, 5]])

# Select columns 0 and 2
torch.index_select(x, dim=1, index=torch.tensor([0, 2]))
# tensor([[0, 2],
#         [3, 5]])
```

**Key Points:**
- `dim=0` selects rows.
- `dim=1` selects columns.
- Indices can be non-consecutive and reordered.
- The selected dimension is preserved.

**Difference from Slicing:**
- Slicing (`x[:, :2]`) selects a continuous range.
- `index_select()` selects specific positions using an index tensor.



````python
indices = torch.LongTensor([0,2])
print(indices)
````

````python
torch.index_select(x,dim =1, index=indices)
````

`torch.index_select()` requires the index tensor to have an integer data type (`int32` or `int64`).
**Key Difference:**
- `torch.Tensor([0, 1])` → `float32` by default.
- `torch.tensor([0, 1])` → `int64` (inferred from input).
- `torch.LongTensor([0, 1])` → `int64`.

````python
torch.index_select(x,dim =1, index = torch.LongTensor([0,1]))
````

````python
torch.index_select(x,dim=0, index=torch.tensor([1]))
````

````python
row_indices = torch.arange(2).long() # [0,1]
print(row_indices)
col_indices = torch.LongTensor([0,1])
print(col_indices)
sub_x = x[row_indices,col_indices] # [[0,1],[3,4]]
````

````python
print(sub_x)
````

## 3. Concatenating Tensors

````python
import torch
x = torch.arange(6).view(2,3)
print(x)
````

````python
x_cat = torch.cat([x,x])
print(x_cat)
````

````python
x_cat = torch.cat([x,x],dim=0)
print(x_cat)
````

````python
x_cat = torch.cat([x,x],dim=1)
print(x_cat)
````

````python
x_stack = torch.stack([x,x])
print(x_stack)
````
