# Activation Functions

## 1. Sigmoid function

```python
import torch
import matplotlib.pyplot as plt

x = torch.arange(-5.,5.,0.1)
print(x)
```

```python
y = torch.sigmoid(x)
print(y)
```

```python
plt.plot(x.numpy(), y.numpy())
plt.show()
```

## 2. Tanh

```python
x = torch.arange(-5.,5.,0.1)
y = torch.tanh(x)
plt.plot(x.numpy(), y.numpy())
plt.show()
```

## 3. ReLU

```python
x = torch.arange(-5.,5.,0.1)
relu = torch.nn.ReLU()
y = relu(x)
plt.plot(x.numpy(),y.numpy())
plt.show()
```

The clipping effect of ReLU that helps with the vanishing gradient problem can also become an issue, where over time certain output in the network can simply become zero and never revive again. This is called the "dying ReLU" problem. To mitigate that effect, variants such as the Leaky ReLU and Parametric ReLU activation functions have proposed, where the leak coefficient _a_ is a learned parameter. 

## 4. PReLU activation function

```python
import torch
import matplotlib.pyplot as plt

prelu = torch.nn.PReLU(num_parameters=1)
x = torch.arange(-5.,5.,0.1)
y = prelu(x)

plt.plot(x.numpy(),y.detach().numpy())
plt.show()
```

## 5. Softmax

```python
import torch.nn as nn
import matplotlib.pyplot as plt

softmax = nn.Softmax(dim=-1)
x_input = torch.randn(1,3) #torch.randn(1, 3) 的意思是：生成一个形状为 (1, 3) 的 Tensor，其中的数值随机取自标准正态分布（Standard Normal Distribution）
print(x_input)
```

```python
y_output = softmax(x_input)
print(y_output)
```
