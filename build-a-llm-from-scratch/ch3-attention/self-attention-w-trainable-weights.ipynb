```python
import torch

inputs = torch.tensor(
    [[0.43,0.15,0.89], # Your (x^1)
     [0.55,0.87,0.66], # journey (x^2)
     [0.57,0.85,0.64], # starts (x^3)
     [0.22,0.58,0.33], # with (x^4)
     [0.77,0.25,0.10], # one (x^5)
     [0.05,0.80,0.55] # step (x^6)
    ]
)

```

```python
x_2 = inputs[1]
d_in = inputs.shape[1] # input dimension 这里是说embedding 后的 input dimension
d_out = 2 # K/V/Q的 那个向量的长度
```

```python
torch.manual_seed(123)
W_query = torch.nn.Parameter(torch.rand(d_in,d_out), requires_grad=False) # 这里暂时使用False 在训练时候会使用True
W_key = torch.nn.Parameter(torch.rand(d_in,d_out), requires_grad=False)
W_value = torch.nn.Parameter(torch.rand(d_in, d_out), requires_grad=False)
```

```python
query_2 = x_2 @ W_query
value_2 = x_2 @ W_value
key_2 = x_2 @ W_key
print("Query, value, key:",query_2,value_2,key_2)
```
